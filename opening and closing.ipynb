```python
# Import the necessary packages
import cv2
import numpy as np
from matplotlib import pyplot as plt
     
```

```python
# Create the text using cv2.putText

img1 = np.zeros((100,400), dtype='uint8')

font = cv2.FONT_HERSHEY_SIMPLEX

cv2.putText(img1,
            'TheAILearner',
            (5,70),
            font,
            2,
            (255),
            5,
            cv2.LINE_AA)

plt.imshow(img1, cmap='gray')
plt.title("Original")
plt.axis('off')
plt.show()
```

```python
# Create the structuring element
kernel1 = cv2.getStructuringElement(
    cv2.MORPH_CROSS,
    (11,11)
)

print(kernel1)
```

```python
# Use the Opening Operation
image_open = cv2.morphologyEx(
    img1,
    cv2.MORPH_OPEN,
    kernel1
)

plt.imshow(image_open, cmap='gray')
plt.title("Opening")
plt.axis('off')
plt.show()
```

```python
# Use the Closing Operation
image_close = cv2.morphologyEx(
    img1,
    cv2.MORPH_CLOSE,
    kernel1
)

plt.imshow(image_close, cmap='gray')
plt.title("Closing")
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.subplot(311)
plt.imshow(img1, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(312)
plt.imshow(image_open, cmap='gray')
plt.title("Opening")
plt.axis('off')

plt.subplot(313)
plt.imshow(image_close, cmap='gray')
plt.title("Closing")
plt.axis('off')

plt.tight_layout()
plt.show()
```
