```python
# Import the necessary packages
import cv2
import numpy as np
from matplotlib import pyplot as plt
     
```

```python
# Create the text using cv2.putText
image = np.zeros((100,400), dtype = 'uint8')
font = cv2. FONT_HERSHEY_SIMPLEX
cv2.putText(image, 'TheAILearner',(5,70), font, 2, (255), 5, cv2. LINE_AA)
```

```python
# Create the structuring element
kernel1 = cv2.getStructuringElement(cv2.MORPH_CROSS, (11,11))
```

```python
# Use the Opening Operation
image1 = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel1)
```

```python
# Use the Closing Operation
image2 = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel1)
```

```python
titles = ['Original Image', 'Opening', 'Closing']

images = [image, image1, image2]

for i in range(3):
    plt.figure(figsize=(8,8))
    
    plt.title(titles[i])
    
    if i == 0:
        plt.imshow(images[i])
    else:
        plt.imshow(images[i], cmap='gray')
    
    plt.axis('off')
    plt.show()
```
